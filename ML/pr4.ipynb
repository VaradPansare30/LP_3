```python
import matplotlib.pyplot as plt
```

```python
# Define function
def func(x):
    return (x + 3) ** 2

# Define derivative
def grad(x):
    return 2 * (x + 3)
```

```python
# Parameters
x = 2
learning_rate = 0.1
epochs = 50

# Store values for plotting
x_vals = [x]
y_vals = [func(x)]
```

```python
for i in range(epochs):
    dx = grad(x)
    x = x - learning_rate * dx

    x_vals.append(x)
    y_vals.append(func(x))

    print("Iteration:", i+1, "x:", x, "f(x):", round(func(x), 5))
```

```python
print("Local minimum at x:", x, "f(x):", round(func(x), 5))
```

```python
# Plot graph
x_plot = list(range(-10, 5))
y_plot = [func(i) for i in x_plot]

plt.plot(x_plot, y_plot, label="y = (x + 3)^2")
plt.scatter(x_vals, y_vals, color='red', label="Gradient Descent Steps")
plt.plot(x_vals, y_vals, '--', color='gray')

plt.title("Gradient Descent")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```
